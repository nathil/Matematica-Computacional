# **Predict** - *Conceito*

Para começarmos a por em prática o conceito de gradiente descendente precisamos compreender o que é o **Predict**. O predict é, basicamente, uma previsão que fazemos incialmente, ou seja, é o nosso chute inicial para a rede neural. Desse modo, chutamos um peso e um bias e se esse chute está certo ou não vai depender do valor da perda que estamos obtendo.
<br/><br/> Podemos definir o predict como uma função, da qual iremos indicar os valores dos pesos, da nossa entrada x, e também do valor constante, que chamaremos de bias. Desse modo, teremos a seguinte função:

$Predict = x_i*w_i + b$

E é essa fórmula que usaremos para fazer um chute do valor y de determinado ponto.

# **Loss** - *Conceito*

O **Loss** é uma função que tem como objetivo medir a diferença entre o resultado previsto pelo modelo e o resultado real. Deste modo, o objetivo que temos com o treinamento é minimizar essa perda/diferença.
<br/> Essa perda é considerada uma peça fundamental no processo de otimização da rede neural, pois ela está fornecendo o quão bem o modelo está atuando em uma determinada tarefa. Sendo assim, a rede neural vai usar o valor da perda para ajustar os seus pesos e outros parâmetros até que a perda seja próxima de 0, ou exatamente 0, o que é muitas vezes quase impossível, por isso ajustamos para valores mais próximos de 0.
<br/> <br/> A função de perda pode ser descrita como:

$\sum_{i=1}^{n} (y_i - (x_i*w+b))^2$

Deste modo, nós podemos derivar essa função para obter o slope/peso de cada valor do intercept (bias), e para isso usamos a **regra da cadeia**, que é uma fórmula que possui como objetivo calcular a derivada de funções compostas.
<br/><br/> Então teremos:
<br/> $(y_i - (x_i*w+b))^2$, que a função que precisamos derivar.
<br/><br/> Transformamos $y_i - (x_i*w+b)$ = u. Logo, $(u)^2$. Aplicando a derivada temos:
<br/> $f'(u) = 2(u)du$, que é a derivada de fora.
<br/> E temos $0 - (0 + b)$, e assim adquirimos: $(-1)$. Que é a derivada de dentro, já que a derivada é em função do intercept, logo, ele é uma determinada variável e os outros valores são constantes, desse modo a derivada do intercept é 1, e como resta apenas o sinal negativo teremos o $(-1)$.  
<br/> Portanto, seguindo com a regra da cadeia: $f'(u)*u$
<br/> $2(u) * -(1)$ = $-2(u)$ = $-2(y_i - (x_i*w_i + b))$
<br/><br/> Então o resultado final da derivada de $(y_i - (x_i*w_i + b))^2$ é $-2(y_i - (x_i*w_i + b))$

# **Learn** - *Conceito*

O aprendizado é uma função da qual iremos atualizar os valores, ou seja, após definirmos o predict e calcularmos o loss usamos a função de aprendizado para ajustar o peso e bias para que os valores sejam o mais próximo para que a perda seja 0. Nessa função, utilizamos também um valor máximo de repetições e o loss rate que vão ajudar a chegarmos no valor esperado.

### *Letra* - **A**

Nesta primeira questão temos como objetivo construir um gradiente descendente, e com isso surge a pergunta: O que um gradiente descendente?

### **O que é gradiente descendente?**

Gradiente descendente é um algoritmo essencial para a otimização do aprendizado de máquina, e usamos ele para diminuir o erro de um determinado modelo do qual o algoritmo melhora ajustando os valores (aprendizado), como já visto anteriormente. E esse processo é feito de modo iterativo no qual a cada passo o algoritmo cálcula uma derivada parcial, ou gradiente, que vai mostrar para onde a direção para qual o algoritmo deve seguir (sempre a contrária de onde está ocorrendo um rápido aumento), isto ocorre para melhorar a precisão do modelo. Esse processo vai durar até que a função esteja próxima ou igual a zero para obter o menor erro que conseguir.

Tendo entendido isso, partimos para a construção do nosso algoritmo.
<br/> Para isso, precisamos da nossa função que cálcula o predict (chute/previsão), deste modo, temos a seguinte função, que é dada com base na fórmula do predict já citada:

$Predict = x_i*w_i + b$

````python
data = [(0.5, 1.4), (2.3, 1.9), (2.9, 3.2)] #dados

#Calculando o predict
def predict(x, intercept, slope):
  return (slope*x) + intercept
````

Depois disso, precisamos da nossa função que cálcula a perda, com isso precisamos de uma váriavel que armazene a soma do quadrado da perda (ou soma residual) e também a soma das derivadas do quadrado, a derivada calculada para essa questão é em relação ao intercept/bias, então temos:
<br/><br/> soma do quadrado da perda:

$\sum_{i=1}^{n} (y_i - (x_i*w+b))^2$

Soma das derivadas:

$\sum_{i=1}^{n} -2(y_i - (x_i*w+b))$

Assim, temos o seguinte código da perda:

````python
#Calculando a perda
def loss(data, intercept, slope):
  result_loss = 0
  result_residual = 0

  for i in range(len(data)):
    result_loss += -2*(data[i][1] - predict(data[i][0], intercept, slope))
    result_residual += (data[i][1] - predict(data[i][0], intercept, slope))**2

  return result_loss, result_residual
````

Após isso, iremos calcular o aprendizado, do qual vamos incluir um valor de repetições que o algoritmo precisará fazer, e também o loss rate para ajudar a calcular o peso conforme o valor do intercept muda.

````python

# Calculando o aprendizado
def learn(data, max_reps, L_rate, intercept, slope):
  list_intercept = []

  for i in range(max_reps):
    result_loss = loss(data, intercept, slope)[0]
    step = result_loss * L_rate
    intercept -= step # Atualizando intercept
    list_intercept.append(intercept)

    if(abs(step) <= 0.0001):
      break

  return intercept, slope, list_intercept
````

Tendo essas funções em mãos, agora podemos construir o gráfico para analisar os resultados e as mudanças que ocorreram.

### **Função Final** - *Gradiente Descendente*

````python
fig, ax = plot()

list_intercept = learn(data, 100, 0.01, 0, 0.64)[2] #lista de cada atualização do bias (intercept)

f_intercept = np.linspace(-2, 3, 100) # 100 pontos entre 0 e 3
f_loss = list(map(lambda intercept: loss(data, intercept, 0.64)[1], f_intercept))

plt.plot(f_intercept, f_loss, color= '#707680') #desenhando a parábola

tangent_line, = plt.plot([-2,3], [0,0], color= '#5287e3') #desenhando a reta tangente

def update(intercept):
  result_loss, result_residual = loss(data, intercept, 0.64) # recebendo soma dos quadrados dos residuos, soma das derivadas
  ax.plot(intercept, result_residual, marker='o', color='red') # desenha os pontos

  tangent = lambda x: result_loss*(x-intercept) + result_residual # cálculo do coenficiente angular
  tangent_line.set_ydata([tangent(-2), tangent(3)]) # calculando a reta a partir do coeficiente e setando os dados para tangent line

  return tangent_line, # retornamos a reta

ani = animation.FuncAnimation(fig=fig, func=update, frames=list_intercept, interval=100)
plt.close()
HTML(ani.to_html5_video())
````

### *Letra* - **B**

O procedimento para a questão B é quase o mesmo para A, no entanto nosso foco será no slope/peso. Ou seja, as derivadas serão feitas em relação ao slope e não mais em relação ao intercept/bias. Sendo assim, as funções também precisarão ser alteradas.

A função para cálcular o predict nesta questão permanece a mesma, desse modo, basta chamar a função para calcular.

Já com relação a função loss, precisaremos calcular uma nova derivada em função do slope, com isso teremos então o $w_i$ será a nossa váriavel principal, enquanto as outras serão apenas valores constantes.
<br/> Desse modo temos:
$\sum_{i=1}^{n} (y_i - (x_i*w_i+b))^2$
<br/><br/> Calculando a derivada obtemos: $\sum_{i=1}^{n} (-2*x_i)*(y_i - (x_i*w_i+b))^2$
<br/><br/> A soma soma dos quadrados continuará a mesma, já que ela é feita antes de calcularmos a derivada.

Assim, a nova função para calcular a perda será:

````python
#calculando a perda com a derivada em função do slope
def loss_slope(data, intercept, slope, **kwargs):
  result_loss_slope = 0
  result_residual = 0

  for i in range(len(data)):
    result_loss_slope += (-2*data[i][0])*(data[i][1] - predict(data[i][0], intercept, slope))
    result_residual += (data[i][1] - predict(data[i][0], intercept, slope))**2

  return None, result_loss_slope, result_residual
````

Para a função de aprendizado, precisaremos adptar também para guardar os valores que variam do peso e também do intercept. Precisamos guardar esses valores, pois, estamos calculando a derivada em função do slope e para observar essa mudança é necessário seguir esse processo.

Portanto, teremos a função:

````python
# Calculando o aprendizado com a derivada em função do slope
def learn_slope(data, max_reps, L_rate, intercept, slope, **kwargs):
  list_slope = []
  list_intercept = []

  for i in range(max_reps):
    # Atualizando o valor do slope
    result_loss_slope = loss_slope(data, intercept, slope)[1]
    step_slope = result_loss_slope * L_rate
    slope -= step_slope
    list_slope.append(slope)

    # Atualizando o valor do intercept
    result_loss_intercep = loss(data, intercept, slope)[0]
    step_intercept = result_loss_intercep * L_rate
    intercept -= step_intercept
    list_intercept.append(intercept)

    if(abs(step_slope) <= 0.0001 and abs(step_intercept) <= 0.0001):
      break

  return intercept, slope, list_slope, list_intercept
````

Assim, teremos o nosso gráfico:

````python
def plot_training(learn_function, loss_function, samples=2):
  fig, ax = plot()

  for point in data:
    ax.plot(point[0], point[1], marker='o', color='green') #desenhando os pontos que temos em data

  _, _, list_slope, list_intercept = learn_function(data, 100, 0.01, 0, 1, samples=samples) #lista de cada atualização do slope (slope) e do bias (intercept)

  predict_line, = plt.plot([-2,3], [0,0], color= '#5287e3')

  def update(frame_data):
    slope, intercept = frame_data # unpacking os valores da tupla
    _, result_loss_slope, result_residual = loss_function(data, intercept, slope, samples=samples) # unpacking os valores calculados em loss_slope

    predict_func = lambda x: predict(x, intercept, slope)
    predict_line.set_ydata([predict_func(-2), predict_func(3)])

    return predict_line,

  ani = animation.FuncAnimation(fig=fig, func=update, frames=zip(list_slope, list_intercept), interval=100)
  plt.close()

  return HTML(ani.to_html5_video())
````

Após isso, precisamos fazer uma versão minibatch desse algoritmo, ou seja, precisaremos adptar mais uma vez nossas funções utilizadas anteriormente para atender as nossas novas necessidades.  

A função loss irá ser atualizada, agora precisameremos do cálculo da derivada em relação ao intercept e ao slope, logo, a função será adptada para calcular as duas ao mesmo tempo, junto também da soma residual:

````python
#calculando a perda com a derivada em função do slope
def loss_minibatch(data, intercept, slope, samples):

  #inicializando variáveis
  result_loss_slope = 0
  result_residual = 0
  result_loss_intercept = 0

  for i in np.random.choice(range(len(data)), size=samples):
    result_loss_slope += (-2*data[i][0])*(data[i][1] - predict(data[i][0], intercept, slope)) # calculando a soma das derivadas em relação ao slope
    result_loss_intercept += -2*(data[i][1] - predict(data[i][0], intercept, slope)) # calculando a soma das derivadas em relação ao intercept
    result_residual += (data[i][1] - predict(data[i][0], intercept, slope))**2 # calculando a soma residual

  return result_loss_slope, result_loss_intercept, result_residual
````

A função de aprendizado não mudará em relação a anterior, então apenas definimos e alteramos os nomes:

````python
# Calculando o aprendizado com a derivada em função do slope
def learn_minibetch(data, max_reps, L_rate, intercept, slope, samples=2):
  list_slope = []
  list_intercept = []

  for i in range(max_reps):
    # Atualizando o valor do slope
    result_loss_slope = loss_minibatch(data, intercept, slope, samples)[0]
    step_slope = result_loss_slope * L_rate
    slope -= step_slope
    list_slope.append(slope)

    # Atualizando o valor do intercept
    result_loss_intercept = loss_minibatch(data, intercept, slope, samples)[1]
    step_intercept = result_loss_intercept * L_rate
    intercept -= step_intercept
    list_intercept.append(intercept)

    if(abs(step_slope) <= 0.0001 and abs(step_intercept) <= 0.0001):
      break

  return intercept, slope, list_slope, list_intercept
````

### *Letra* - **D**

Para a questão D, iremos agora construir uma rede neural, portanto, precisaremos agora utilizar vetores e matriz para fazer nossa aplicação.

A nossa função predict terá como base os valores de entrada e saída, para calculá-lo usaremos a mesma ideia base $ x_i*w_i + b$, porém agora com vetores e matrizes.
<br/><br/> Primeiramente, calculamos o resultado da camada hidden com a função citada acima, recebemos o vetor de entradas e multiplicamos com a matriz dos pesos da camada hidden, após isso somamos com o vetor dos bias da camada hidden, e após esse processo iremos aplicar a função de ativação nesse resultado do qual iremos adquirir um novo vetor.
<br/> Calculando o resultado da camada hidden:
```
(hidden_weight @ vet_input) + hidden_bias
```
<br/> Aplicando a função de ativação:
```
softplus((hidden_weight @ vet_input) + hidden_bias)
```
<br/>Com isso, multiplicamos o novo vetor pela matriz de pesos de saída e somamos com os vetores de bias da saída também e assim obtemos o nosso predict que é um vetor de apenas uma dimensão.

<br/> Calculando o resultado da camada de saída
```
(output_weight @ result_hidden) + output_bias
```

<br/> A nossa função de ativação softplus, é dada pelo seguinte cálculo:
$log(1+e^x)$

<br/> Essa função é conhecida por ser uma função suave e diferenciável, funcionando como uma aproximação "leve" da ReLU. Ela é suave e continuamente diferenciável em todos os pontos, o que é bom para determinados tipos de otimizações.

Deste modo, teremos o código para o predict:

````python
points = [(0,0), (0.5, 1), (1,0)] # pontos especificados no material

def predict_neural_network(input, b_3=0):
  vet_input = np.array([input]) # vetor com os inputs

  hidden_weight = np.array([[3.34], [-3.53]]) # matriz com os pesos da camada hidden
  hidden_bias = np.array([-1.43, 0.57] ) # vetor com os bias da camada hidden

  result_hidden = softplus((hidden_weight @ vet_input) + hidden_bias) # vetor resultante com a função de ativação

  output_weight = np.array([[-1.22, -2.30]])
  output_bias = np.array([b_3])

  result_output = (output_weight @ result_hidden) + output_bias

  return result_output[0]

print(predict_neural_network(0))
````

Código para a função de ativação:

````python
def softplus(array):
  return np.log(1 + np.exp(array))
````

Para a função loss seguiremos a mesma ideia inicial para construir igual a função loss da letra A, que será uma derivada apenas em função do bias, teremos então:

````python
#calculando a perda com a derivada em função do bias
def loss_neural_network(data, b_3):
  return sum(-2*(data[i][1] - predict_neural_network(data[i][0], b_3)) for i in range(len(data)))
````

E para a função de aprendizado, guardaremos apenas o valor do bias também, já que é o fizemos a derivada em função:

````python
# Calculando o aprendizado
def learn_neural_network(data, max_reps, L_rate, b_3):
  list_b_3 = []

  for i in range(max_reps):
    result_loss = loss_neural_network(data, b_3)
    old_b_3 = b_3
    step = result_loss * L_rate
    b_3 -= step # Atualizando b_3
    list_b_3.append(b_3)

    print(f'Step Size: {step}, New B_3: {b_3}, Old B_3: {old_b_3}')

    if(abs(step) <= 0.0001):
      break

  return b_3, list_b_3
````

Deste modo, podemos plotar o gráfico para observar o resultado:

````python
# plotando o gráfico
fig, ax = plot()

plt.axhline(0)
plt.axvline(0)

for point in points:
    ax.plot(point[0], point[1], marker='o', color='green') #desenhando os pontos que temos em points

ax.set_ylim(-3,3)

f_x = np.linspace(-2, 3, 100) # 100 pontos entre 0 e 3
f_neural_network = list(map(lambda x: predict_neural_network(x), f_x))

_, list_b_3 = learn_neural_network(points, 100, 0.01, 0)

predict_line, = plt.plot(f_x, f_neural_network, color= '#707680') #desenhando a parábola

def update(b_3):
  predict_func = lambda x: predict_neural_network(x, b_3)
  predict_line.set_ydata(list(map(predict_func, f_x)))

  return predict_line,

ani = animation.FuncAnimation(fig=fig, func=update, frames=list_b_3, interval=100)
plt.close()
HTML(ani.to_html5_video())
````

### *Letra* - **E**

Backpropagation é uma técnica de aprendizado de máquina de extrema importância para a otimização de redes neurais. Que permite a utilização de algoritmos de gradiente descendente para atualizar os pesos, que é como os modelos aprendem.

Sendo assim, precisaremos calcular um novo predict backpropagation. Pois, teremos de utilizar a derivada em função de $b_3$, $w_3$, $w_4$, e ainda utilizaremos a função de ativação softplus. Deste modo, a função se dá por:

````python
points = [(0,0), (0.5, 1), (1,0)]

def softplus(array):
  return np.log(1 + np.exp(array))

def predict_backpropagation(input, b_3=0, w3=1, w4=1):
  vet_input = np.array([input]) # vetor com os inputs

  hidden_weight = np.array([[3.34], [-3.53]]) # matriz com os pesos da camada hidden
  hidden_bias = np.array([-1.43, 0.57] ) # vetor com os bias da camada hidden

  result_hidden = softplus((hidden_weight @ vet_input) + hidden_bias) # vetor resultante com a função de ativação

  output_weight = np.array([[w3, w4]])
  output_bias = np.array([b_3])

  result_output = (output_weight @ result_hidden) + output_bias

  return result_output[0], result_hidden[0], result_hidden[1]

````

A nossa função loss será feita com base também nas derivadas do fatores $b_3$, $w_3$, $w_4$. A função será:

````python
#calculando a perda com a derivada em função do bias
def loss_backpropagation(data, b_3, w3, w4):
  result_loss_b3 = 0
  result_loss_w3 = 0
  result_loss_w4 = 0
  for i in range(len(data)):
    result_output, result_hidden_0, result_hidden_1 = predict_backpropagation(data[i][0], b_3, w3, w4)

    result_loss_b3 += -2*(data[i][1] - result_output)
    result_loss_w3 += (-2*result_hidden_0)*(data[i][1] - result_output)
    result_loss_w4 += (-2*result_hidden_1)*(data[i][1] - result_output)

  return  result_loss_b3, result_loss_w3, result_loss_w4
````

Na função learn precisaremos adicionar uma lista para receber as atualizações dos três valores:

````python
# Calculando o aprendizado
def learn_backpropagation(data, max_reps, L_rate, b_3, w3, w4):
  list_b_3 = []
  list_w3 = []
  list_w4 = []

  for i in range(max_reps):
    result_loss_b3, result_loss_w3, result_loss_w4 = loss_backpropagation(data, b_3, w3, w4)
    old_b_3 = b_3
    old_w3 = w3
    old_w4 = w4

    step_b3 = result_loss_b3 * L_rate
    step_w3 = result_loss_w3 * L_rate
    step_w4 = result_loss_w4 * L_rate

    b_3 -= step_b3 # Atualizando b_3
    w3 -= step_w3 # Atualizando w3
    w4 -= step_w4 # Atualizando w4

    list_b_3.append(b_3)
    list_w3.append(w3)
    list_w4.append(w4)

    print(f'Step Size B_3: {step_b3}, New B_3: {b_3}, Old B_3: {old_b_3}')
    print(f'Step Size W3: {step_w3}, New W3: {w3}, Old W3: {old_w3}')
    print(f'Step Size W4: {step_w4}, New W4: {w4}, Old W4: {old_w4}')

    if(abs(step_b3) <= 0.0001 and abs(step_w3) <= 0.0001 and abs(step_w4) <= 0.0001):
      break

  return b_3, w3, w4, list_b_3, list_w3, list_w4
````
